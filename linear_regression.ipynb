```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv("smart_class_energy_data.csv")
df.head()
df.describe()
```

```python
df.dropna()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
df["Time_of_Day"] = le.fit_transform(df["Time_of_Day"])
```

```python
correlation_matrix = df.corr()
print(correlation_matrix)
```

```python
plt.figure(figsize=(10,6)) # Adjust figure size
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f",linewidths=0.5)
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
def remove_outliers_iqr(df):
    Q1 = df.quantile(0.25)
    Q3 = df.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return df[~((df < lower_bound) | (df > upper_bound)).any(axis=1)]


df_cleaned = remove_outliers_iqr(df)

print(df_cleaned)
```

```python
correlation_matrix = df_cleaned.corr()
```

```python

# Calculate correlation matrix after outlier removal
correlation_matrix = df_cleaned.corr()

# Plot heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix Heatmap (After Outlier Removal)")
plt.show()
```

```python
# Plot Original and Cleaned Data
plt.figure(figsize=(10, 5))

# Plot Original Data
plt.plot(
    df.index,
    df.mean(axis=1),
    label="Original Data",
    linestyle="dashed",
    color="red"
)

# Plot Cleaned Data
plt.plot(
    df_cleaned.index,
    df_cleaned.mean(axis=1),
    label="Without Outliers",
    color="blue"
)

# Labels and Title
plt.xlabel("Index")
plt.ylabel("Mean Value of Rows")
plt.title("Line Graph Before and After Removing Outliers")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
x=df.drop(columns=['AC_Temp'])
```

```python
y=df['AC_Temp']
```

```python
y
```

```python
x_train, x_test, y_train,y_test=train_test_split(x,y,test_size=20,random_state=40)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
mse=mean_squared_error(y_test,y_pred)
```

```python
print(mse)
```

```python
r2=r2_score(y_test,y_pred)
```

```python
print(r2)
```
